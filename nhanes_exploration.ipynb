Dataset Report: Glycohemoglobin (GHB_L)
Data File: GHB_L.xpt

I looked at the August 2021-2023 NHANES glycohemoglobin file (GHB_L.xpt). It contains 7199 rows (participant records) and 3 columns (“SEQN,” “WTPH2YR,” and “LBXGH.” “SEQN” is the sequence number that identifies each participant, “WTPH2YR”  is the phlebotomy 2 year weight that was included to address possible nonresponsive bias, and “LBXGH” is the percentage of glycohemoglobin. For our research project, “SEQN” would not be necessary for our model, as we do not need patient IDs, since patient IDs are not pieces of health information that could help predict what test the patient needs to have next. However, I think “WTPH2YR” could be helpful, as it is a weight that helps counteract biases in our data. If we were to create a model for a disease that would require glycohemoglobin percentages, “LBXGH” would be a helpful feature.
484 records are missing LBXGH, leaving 6715 records (counts). Since there are no columns that provide a diagnosis or other outcome label, this file has no classes to report. 25% of the participants had 5.2% and below of glucose attached to their hemoglobin, while 50% had 5.5% and below of glucose attached to their hemoglobin, and 75% had 5.8% and below of glucose attached to their hemoglobin.
