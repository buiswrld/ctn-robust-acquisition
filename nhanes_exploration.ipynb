I looked at the August 2021-2023 NHANES glycohemoglobin file (GHB_L.xpt). It contains 7199 rows (participant records) and 3 columns (“SEQN,” “WTPH2YR,” and “LBXGH"). “SEQN” is the sequence number that identifies each participant, “WTPH2YR”  is the phlebotomy 2 year weight that was included to address possible nonresponsive bias, and “LBXGH” is the percentage of glycohemoglobin. 484 records are missing LBXGH, leaving 6715 records (counts). Since there are no columns that provide a diagnosis or other outcome label, this file has no classes to report. 25% of the participants had 5.2% and below of glucose attached to their hemoglobin, while 50% had 5.5% and below of glucose attached to their hemoglobin, and 75% had 5.8% and below of glucose attached to their hemoglobin. The mean of these records is “5.709471,” std is “1.054899,” max is “17.100000,” min is “3.200000.
